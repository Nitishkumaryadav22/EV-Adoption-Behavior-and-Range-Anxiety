# Handling Missing Values

```python
missdf = pd.DataFrame()
```

```python
missdf['columns']=df.columns
missdf['normalize']=nor.values
```

```python
missdf
```

```python
miss = missdf[missdf['normalize']>0]
```

```python
for i in miss['columns']:
    if df[i].dtype == 'int64' or df[i].dtype == 'float64':
        df[i] = df[i].fillna(df[i].mean())
    else:
        df[i] = df[i].fillna(df[i].mode()[0])
```

```python
df.isna().sum()
```

# Detecting And Treating Outliers

```python
Q1 = df[numerical_columns].quantile(0.25)
Q3 = df[numerical_columns].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
```

```python
outlier_count = ((df[numerical_columns] < lower_limit) | 
                 (df[numerical_columns] > upper_limit)).sum()

outlier_count
```

```python
from feature_engine.outliers import Winsorizer
```

```python
Win = Winsorizer(capping_method='iqr',tail='both')
```

```python
df = Win.fit_transform(df)
```
